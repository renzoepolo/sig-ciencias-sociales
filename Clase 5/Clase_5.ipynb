## 4. Qué es la cartografía temática

Hay dos grandes familias de mapas.

Un **mapa base** (o topográfico) describe el territorio: ríos, rutas, localidades, curvas de
nivel. Responde *¿qué hay acá?*. Es lo que veníamos usando como fondo o como referencia.

Un **mapa temático** representa una variable sobre el territorio. Responde *¿cómo se
distribuye esto en el espacio?*. El mapa base aparece solo para que el lector
se ubique.

La cartografía temática es una disciplina anterior a los SIG, con reglas construidas a lo
largo de un siglo y con evidencia experimental sobre cómo la gente lee un mapa.

### 4.1 Las variables visuales

Jacques Bertin, en *Sémiologie graphique* (1967), mostró que todo signo dibujado en un mapa
puede variar de **seis** maneras. Son las **variables visuales**:

| Variable visual | En qué consiste | Qué transmite bien |
|---|---|---|
| **Tamaño** | Un símbolo más grande o más chico | Cantidades. Es la única que se lee como "más" o "menos" de forma inmediata |
| **Valor** | Más claro o más oscuro | Orden: de poco a mucho |
| **Grano o textura** | Trama más fina o más gruesa | Orden, cuando no se puede usar color |
| **Color (tono)** | Rojo, azul, verde | Categorías **sin** orden: tipos de suelo, partidos políticos |
| **Orientación** | Inclinación de la trama | Categorías sin orden |
| **Forma** | Círculo, triángulo, cuadrado | Categorías sin orden |

📖 *Diseño Cartográfico*, cap. II y IV, desarrollan las seis variables y qué percibe el ojo
de cada una.

<img src="https://raw.githubusercontent.com/renzoepolo/sig-ciencias-sociales/main/Clase%205/variables_visuales.png" width="900">

**Seis mapas de Andalucía, uno por variable visual.** Arriba: el **tamaño** de los
círculos, la **textura** de las tramas y el **tamaño** aplicado a líneas —el grosor del
trazo—. Abajo: el **color** de las líneas, el **valor** —claro a oscuro— de las líneas, y
el mismo valor aplicado a las superficies, que es el mapa de coropletas.

Fijate en la diferencia entre los dos de abajo a la izquierda: las líneas de colores
distinguen **tipos** de ruta; las líneas en grises distinguen **cantidades**. Es el mismo
dibujo con una variable visual distinta, y dice cosas de naturaleza diferente.

### 4.2 Qué mide la variable: el nivel de medición

Antes de elegir el mapa hay que saber qué clase de número tenemos. La distinción viene de la
estadística y es la misma que se usa en cualquier materia de metodología:

| Nivel | Ejemplo |
|---|---|
| **Nominal** | Tipo de cobertura del suelo, provincia |
| **Ordinal** | Nivel educativo alcanzado, "bajo / medio / alto" | 
| **Cuantitativo** | Hogares con NBI, superficie, porcentaje |

Y dentro de los cuantitativos, una distinción que va a ser el eje de la clase:

- **Absoluto**: un conteo.
- **Relativo**: una razón, tasa o porcentaje.

📖 *Cartografía Temática*, cap. 2, trata este paso como el primero del proceso cartográfico, antes de dibujar nada.

### 4.3 Los tipos de mapa temático

Con la variable visual y tipo de medición se elige el tipo de mapa:

| Tipo de mapa | Cómo representa | Para qué dato sirve |
|---|---|---|
| **Coropletas** | Pinta cada unidad areal según su valor (variable visual: **valor**) | Datos **relativos** por unidad areal: porcentajes, tasas, densidades |
| **Símbolos proporcionales** | Un círculo en cada lugar, más grande cuanto mayor el valor (variable visual: **tamaño**) | Datos **absolutos**: conteos, totales |
| **Mapa de puntos** | Un punto por cada *n* casos, repartidos dentro de la unidad | Datos absolutos, cuando importa la textura de la distribución |
| **Isolíneas** | Líneas que unen puntos de igual valor | Fenómenos **continuos**: temperatura, altura, presión |
| **Mapa de calor** (densidad) | Superficie continua de colores calculada a partir de puntos | **Eventos puntuales individuales**: delitos, llamados, focos de incendio |
| **Cartograma** | Deforma el tamaño de cada unidad según el valor | Cuando el dato importa más que la forma del territorio |
| **Dasimétrico** | Reparte el valor de un área según otra información | Cuando se sabe que el fenómeno no es uniforme adentro de la unidad |

## 6. Un mapa de coropletas, paso a paso

> 📖 *Cartografía Temática*, cap. 6 — Mapas de coropletas.

### 🧭 Concepto

Un **mapa de coropletas** (del griego *khōra*, territorio, y *plēthos*, cantidad) pinta cada
unidad areal según el valor que toma la variable ahí. Cada unidad —acá, cada departamento— se
llama **corograma**.

El supuesto que hace es fuerte y conviene tenerlo presente: **adentro de cada unidad el valor
es constante**. El mapa muestra escalones, no una superficie suave. Si la pobreza varía mucho
dentro de un mismo departamento, el mapa no lo va a mostrar.

Vamos a construir el mapa **agregando un elemento por vez**, para ver qué hace cada uno.

### ▶️ Paso 1 — El mapa más simple que existe

Una sola línea: `.plot()` con el nombre de la columna. GeoPandas le asigna un color a cada
departamento según su valor.

```python
departamentos.plot(column="hogares_nbi")
```

Sale un mapa, pero no sirve para nada: es chico, no sabemos qué significan los colores y los
ejes muestran coordenadas que al lector no le importan. Vamos por partes.

### ▶️ Paso 2 — Tamaño y paleta de colores

Creamos la figura con el tamaño que queremos, y elegimos la **paleta** con `cmap`.

Acá aplicamos la teoría: la variable va de menos a más, así que usamos la variable visual
**valor** —de claro a oscuro—, con una paleta **secuencial**. `"Reds"` es una de ellas. El
catálogo completo está en la
[documentación de Matplotlib](https://matplotlib.org/stable/users/explain/colors/colormaps.html).

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="hogares_nbi", cmap="Reds", ax=eje)

plt.show()
```

### ▶️ Paso 3 — La leyenda

Sin leyenda, un mapa de coropletas no se puede leer. Se agrega con `legend=True`.

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="hogares_nbi", cmap="Reds", legend=True, ax=eje)

plt.show()
```

Apareció una **barra de color continua**: a cada valor entre 5 y 52.709 le corresponde un
tono. Y ahí se ve el problema: casi todo el país quedó del color más claro.

Es la asimetría del histograma, ahora convertida en un problema de dibujo. La barra reparte
los tonos de manera uniforme entre el mínimo y el máximo, y como casi todos los valores están
abajo, casi todos los departamentos reciben el mismo color.

### ▶️ Paso 4 — Agrupar en clases

La solución es no usar una escala continua sino **agrupar los valores en unas pocas clases** y
darle un color a cada una. Eso es lo que hace `scheme`, que GeoPandas le pide a Mapclassify.

Por ahora usamos `"Quantiles"` con 5 clases: cinco grupos con la misma cantidad de
departamentos cada uno. En el bloque 8 discutimos esta elección en serio.

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="hogares_nbi", cmap="Reds",
                   scheme="Quantiles", k=5,
                   legend=True, ax=eje)

plt.show()
```

Ahora sí se ve algo: la leyenda pasó a ser una lista de **rangos** y el mapa dejó de estar
uniformemente claro.

### ▶️ Paso 5 — Limpiar y dar referencia

Tres arreglos de presentación:

- `set_axis_off()` saca los ejes con las coordenadas, que no aportan nada.
- `edgecolor` y `linewidth` dibujan el borde de cada departamento en blanco y fino, para que
  se distingan sin que las líneas tapen el color.
- Encima dibujamos el **contorno de las provincias**, para que el lector se ubique. Es el
  único rol del mapa base en un mapa temático.

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="hogares_nbi", cmap="Reds",
                   scheme="Quantiles", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True, ax=eje)

provincias.boundary.plot(ax=eje, color="grey", linewidth=0.6)

eje.set_axis_off()
plt.show()
```

### ▶️ Paso 6 — Título, leyenda con nombre y fuente

Falta lo que convierte un gráfico en un mapa publicable: decir qué muestra, qué significa la
leyenda y de dónde salió el dato.

`legend_kwds` controla la leyenda: `title` le pone un nombre y `loc` la ubica adentro del
mapa (acá abajo a la derecha, donde hay lugar).

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="hogares_nbi", cmap="Reds",
                   scheme="Quantiles", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True,
                   legend_kwds={"title": "Hogares con NBI", "loc": "lower right"}, #"upper right", "center", "center left" "best", etc
                   ax=eje)

provincias.boundary.plot(ax=eje, color="grey", linewidth=0.6)

eje.set_title("Hogares con al menos un indicador NBI\nArgentina, departamentos, Censo 2022")
eje.set_axis_off()

eje.annotate("Fuente: INDEC, Censo 2022. Clasificación: cuantiles, 5 clases.",
             xy=(0.0, -0.02), xycoords="axes fraction", fontsize=8, color="grey")

plt.show()
```

### 🔍 Interpretación — un mapa terminado, y equivocado

Ese último mapa está bien dibujado: tiene título, leyenda con nombre, fuente y nota
metodológica. Y sin embargo **incumple la primera regla del capítulo 6**: es un mapa de
coropletas de un dato **absoluto**.

Hay departamentos enormes pintados de tonos oscuros. ¿Son pobres? No
necesariamente: son **grandes y poblados**, y por eso acumulan muchos hogares con NBI en
total. El color ocupa una superficie gigante y el ojo le cree a la superficie.

## 7. Qué número va en el mapa: conteo, porcentaje y densidad

> 📖 *Cartografía Temática*, cap. 2 — Procesamiento de los datos.

### 🧭 Concepto

El mismo dato se puede expresar de tres maneras, y cada una
responde una **pregunta distinta**:

| Medida | Cálculo | Pregunta que responde |
|---|---|---|
| **Conteo** | hogares con NBI | ¿Dónde **existen más** hogares con NBI? |
| **Porcentaje** | hogares con NBI ÷ total de hogares × 100 | ¿Dónde es **más probable** que un hogar tenga NBI? |
| **Densidad** | hogares con NBI ÷ km² | ¿Dónde **se concentran** en el espacio? |

Vamos a calcular las dos que faltan y a mapear cada una.

### ▶️ Paso 1 — El porcentaje

Dividimos por el total de hogares del departamento, que es la población de referencia:
cuántos hogares **podrían** haber tenido NBI. Esto se llama **normalizar** la variable.

```python
departamentos["porcentaje_nbi"] = departamentos["hogares_nbi"] / departamentos["hogares"] * 100

departamentos["porcentaje_nbi"].describe().round(1)
```

```python
fig, eje = plt.subplots(figsize=(8, 4))

eje.hist(departamentos["porcentaje_nbi"], bins=40, color="#c0392b")
eje.set_title("Porcentaje de hogares con NBI por departamento — Censo 2022")
eje.set_xlabel("% Hogares con NBI")
eje.set_ylabel("cantidad de departamentos")

plt.show()
```

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="porcentaje_nbi", cmap="Reds",
                   scheme="Quantiles", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True,
                   legend_kwds={"title": "% de hogares con NBI", "loc": "lower right"},
                   ax=eje)

provincias.boundary.plot(ax=eje, color="grey", linewidth=0.6)

eje.set_title("Porcentaje de hogares con NBI\nArgentina, departamentos, Censo 2022")
eje.set_axis_off()
plt.show()
```

El mapa cambió por completo.

### ▶️ Paso 2 — La densidad

Ahora dividimos por la superficie. El resultado ya no es una proporción de hogares sino una
**concentración en el espacio**: cuántos hogares con NBI hay por kilómetro cuadrado.

```python
departamentos["densidad_nbi"] = departamentos["hogares_nbi"] / departamentos["superficie_km2"]

departamentos["densidad_nbi"].describe().round(1)
```

## 8. La clasificación: ¿dónde se cortan las clases?

> 📖 *Cartografía Temática*, cap. 6, apartado 5 — Clasificación de los datos.

### 🧭 Concepto

Ya vimos que un mapa de coropletas no pinta 527 colores distintos: agrupa los departamentos en
unas pocas clases. Toda clasificación **pierde información** a cambio de legibilidad: el lector
deja de ver el valor exacto de cada unidad y ve un patrón general.

Ahí hay dos decisiones: **cuántas clases** y **dónde cortarlas**.

Sobre la cantidad, la práctica cartográfica recomienda **entre 4 y 7**. Con menos, el mapa no
muestra variación; con más, el ojo no distingue los tonos. Vamos a usar 5.

Sobre los cortes, Mapclassify trae muchos métodos. Vemos los tres más usados en ciencias
sociales, **uno por uno y siempre con la misma variable**: el porcentaje de hogares con NBI.

```python
# Todos los métodos que ofrece mapclassify
mc.CLASSIFIERS
```

### ▶️ Método 1 — Intervalos iguales

Divide el rango —del mínimo al máximo— en cinco tramos del **mismo ancho**. Es el más
intuitivo: cada color abarca la misma cantidad de puntos porcentuales.

```python
porcentaje = departamentos["porcentaje_nbi"]

intervalos_iguales = mc.EqualInterval(porcentaje, k=5)
intervalos_iguales
```

La salida muestra los cortes y, en la columna de la derecha, cuántos departamentos caen en cada
clase. Veámoslo sobre el histograma: cada línea roja es un corte.

```python
fig, eje = plt.subplots(figsize=(8, 4))

eje.hist(porcentaje, bins=40, color="lightgrey")
for corte in intervalos_iguales.bins:
    eje.axvline(corte, color="red")

eje.set_title("Intervalos iguales — los cortes sobre la distribución")
eje.set_xlabel("% de hogares con NBI")
plt.show()
```

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="porcentaje_nbi", cmap="Reds",
                   scheme="EqualInterval", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True, legend_kwds={"loc": "lower right"},
                   ax=eje)

eje.set_title("% de hogares con NBI — Intervalos iguales")
eje.set_axis_off()
plt.show()
```

Los cortes están repartidos parejo, pero los datos no. Como la distribución es asimétrica,
**356 departamentos —dos de cada tres— caen en la primera clase** y la última tiene 2. El mapa
muestra un país casi uniforme con un par de focos.

### ▶️ Método 2 — Cuantiles

Al revés: pone la **misma cantidad de departamentos en cada clase**, y para eso corta donde haga
falta. Garantiza que todos los colores aparezcan en el mapa.

```python
cuantiles = mc.Quantiles(porcentaje, k=5)
cuantiles
```

```python
fig, eje = plt.subplots(figsize=(8, 4))

eje.hist(porcentaje, bins=40, color="lightgrey")
for corte in cuantiles.bins:
    eje.axvline(corte, color="red")

eje.set_title("Cuantiles — los cortes sobre la distribución")
eje.set_xlabel("% de hogares con NBI")
plt.show()
```

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="porcentaje_nbi", cmap="Reds",
                   scheme="Quantiles", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True, legend_kwds={"loc": "lower right"},
                   ax=eje)

eje.set_title("% de hogares con NBI — Cuantiles")
eje.set_axis_off()
plt.show()
```

Ahora el mapa tiene variación en todo el país: hay 105 o 106 departamentos de cada color. Pero
mirá los cortes en el histograma: los cuatro primeros están amontonados a la izquierda y el
último abarca un tramo enorme. **La clase más oscura va de 11,3 % a 39,9 %**: un departamento
con 12 % de hogares con NBI se pinta igual que uno con 40 %.

### ▶️ Método 3 — Quiebres naturales (Fisher-Jenks)

Busca los cortes que hacen las clases **lo más homogéneas posible por dentro y lo más distintas
posible entre sí**. En vez de imponer un criterio, se adapta a la forma de la distribución. Es
más lento de calcular, porque prueba muchas combinaciones.

⚠️ En la bibliografía y en muchos programas aparece como *"Natural Breaks (Jenks)"*. Mapclassify
tiene dos implementaciones parecidas pero distintas, `FisherJenks` y `NaturalBreaks`.

```python
quiebres_naturales = mc.FisherJenks(porcentaje, k=5)
quiebres_naturales
```

```python
fig, eje = plt.subplots(figsize=(8, 4))

eje.hist(porcentaje, bins=40, color="lightgrey")
for corte in quiebres_naturales.bins:
    eje.axvline(corte, color="red")

eje.set_title("Quiebres naturales — los cortes sobre la distribución")
eje.set_xlabel("% de hogares con NBI")
plt.show()
```

```python
fig, eje = plt.subplots(figsize=(7, 10))

departamentos.plot(column="porcentaje_nbi", cmap="Reds",
                   scheme="FisherJenks", k=5,
                   edgecolor="white", linewidth=0.1,
                   legend=True, legend_kwds={"loc": "lower right"},
                   ax=eje)

eje.set_title("% de hogares con NBI — Quiebres naturales (Fisher-Jenks)")
eje.set_axis_off()
plt.show()
```

### ▶️ Los tres, uno al lado del otro

Para comparar: los tres mapas, y cuántos departamentos quedaron en cada clase.

```python
metodos = {
    "EqualInterval": "Intervalos iguales",
    "Quantiles": "Cuantiles",
    "FisherJenks": "Quiebres naturales",
}

fig, ejes = plt.subplots(1, 3, figsize=(16, 9))

for eje, metodo in zip(ejes, metodos):
    departamentos.plot(column="porcentaje_nbi", cmap="Reds",
                       scheme=metodo, k=5,
                       edgecolor="white", linewidth=0.1,
                       legend=True, legend_kwds={"loc": "lower right", "fontsize": 8},
                       ax=eje)
    eje.set_title(metodos[metodo])
    eje.set_axis_off()

plt.show()
```

### 🔍 Interpretación

Es el mismo dato, en los mismos departamentos. Solo cambió dónde se corta.

Con **intervalos iguales** el país parece bastante parejo salvo dos casos extremos. Con
**cuantiles**, el norte entero se ve crítico, en una clase que junta departamentos con 12 % y
con 40 %. Con **quiebres naturales** los cortes siguen la forma del histograma: 25 departamentos
en la cuarta clase y los 2 extremos solos en la quinta.

Ninguno está mal hecho. Con el mapa de cuantiles se podría titular *"El norte entero, en
situación crítica"*; con el de intervalos iguales, *"Pocas diferencias, salvo dos
departamentos"*. **El dato es el mismo: la historia la eligió la clasificación.**

Por eso se mira el histograma **antes** de elegir el método, y por eso la leyenda con los cortes
no es un adorno: es la parte del mapa que dice cómo leerlo.

### ⚠️ Tres sesgos de los mapas de coropletas

Para tenerlos presentes siempre que se lee uno, propio o ajeno:

1. **Sesgo de percepción por superficie.** Las unidades grandes llaman más la atención que las
   chicas, aunque el dato sea igual de importante.
2. **Arbitrariedad de las clases.** La elección de los cortes es una decisión de quien hace el
   mapa, y cambia la imagen.
3. **Sensibilidad a cambios chicos.** Un departamento que pasa de 11,2 % a 11,4 % puede cambiar
   de clase y, en el mapa, de color. La variación real es mínima y el salto visual es grande.

## 9. Símbolos proporcionales: los conteos, con círculos

> 📖 *Cartografía Temática*, cap. 3 — Mapas de símbolos proporcionales.

### 🧭 Concepto

Volvamos al problema del bloque 6: el conteo de hogares con NBI es un dato absoluto y no
corresponde pintarlo. ¿Y entonces cómo se mapea un total?

Con **símbolos proporcionales**: un símbolo en cada lugar, más grande cuanto mayor es el valor.
Se usa la variable visual **tamaño**, que es la que el ojo lee directamente como cantidad. El
símbolo habitual es el círculo, porque es compacto y visualmente estable.

Y acá aparece la decisión que la mayoría de los mapas se equivoca. ¿Qué tiene que ser
proporcional al valor: **el radio** del círculo o **su área**?

El capítulo 3 es terminante: *"el círculo es un símbolo superficial por lo que su superficie debe
ser proporcional al valor que representa"*. La razón es perceptiva: el ojo compara
**superficies**. Y como el área de un círculo crece con el **cuadrado** del radio, si duplicamos
el radio para un valor que es el doble, el círculo termina ocupando **cuatro** veces más
superficie. El mapa exagera las diferencias.

En Matplotlib el tamaño de un punto (`markersize`) se expresa como **área**. Así que:

- `markersize = valor` → el área es proporcional al valor ✔️

### 👀 Paso 1 — Dos círculos

Antes del mapa, el caso más simple: un valor de 100 y otro de 400, que es cuatro veces mayor. A
la izquierda escalando el radio; a la derecha, el área.

```python
valores = pd.Series([100, 400])

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.5))

# Izquierda: el radio crece igual que el valor
ejes[0].scatter([0, 1], [0, 0], s=(valores / 10) ** 2, color="#c0392b")
ejes[0].set_title("Radio proporcional al valor ✖️")

# Derecha: el área crece igual que el valor
ejes[1].scatter([0, 1], [0, 0], s=valores, color="#c0392b")
ejes[1].set_title("Área proporcional al valor ✔️")

for eje in ejes:
    eje.set_xlim(-0.6, 1.6)
    eje.set_xticks([0, 1], ["valor 100", "valor 400"])
    eje.set_yticks([])

plt.show()
```

### ▶️ Paso 2 — Un punto por departamento

Para poner un círculo en cada departamento necesitamos **un punto**, no un polígono. Usamos el
**centroide**, que ya conocemos de la Clase 2: la localización que representa al área.

```python
puntos = departamentos.copy()
puntos["geometry"] = departamentos.centroid

puntos[["departamento", "hogares_nbi", "geometry"]].head()
```

### ▶️ Paso 3 — Calcular los tamaños

```python
maximo = puntos["hogares_nbi"].max()
proporcion = puntos["hogares_nbi"] / maximo      # un número entre 0 y 1

circulo_por_area = 600 * proporcion              # el área crece como el valor ✔️
```

### 👀 Paso 4 — El mapa

```python
fig, ejes = plt.subplots(figsize=(14, 10))

provincias.plot(ax=ejes, color="#f0f0f0", edgecolor="grey", linewidth=0.4)
puntos.plot(ax=ejes, markersize=circulo_por_area, color="#c0392b", alpha=0.6)
ejes.set_title("Área proporcional al valor")

eje.set_axis_off()
plt.show()
```

## 10. Mapas interactivos con Folium

### 🧭 Concepto

Todos los mapas de hasta acá son **estáticos**: una imagen. Sirven para un informe impreso, un
artículo o una presentación, y tienen una limitación grande: no se puede saber **qué departamento
es cada polígono**.

Un mapa **interactivo** se abre en el navegador y permite ampliar, desplazarse y consultar cada
elemento. Se usa para **explorar** los datos y para publicarlos en la web.

En Python la biblioteca es **Folium**, que por debajo usa Leaflet, la misma herramienta con la que
están hechos muchos mapas de los diarios. GeoPandas la integra en un solo método: `.explore()`.

### ▶️ Paso 1 (OPCIONAL) — Simplificar las geometrías

Antes de armar el mapa interactivo reducimos la cantidad de vértices con `simplify()`, la función
de generalización de la Clase 2. Con una tolerancia de 500 metros el contorno sigue siendo el
mismo a escala nacional, y el mapa pesa mucho menos.

```python
livianos = departamentos.copy()
livianos["geometry"] = departamentos.simplify(500)   # 500 metros de tolerancia

provincias_livianas = provincias.copy()
provincias_livianas["geometry"] = provincias.simplify(500)

print("Vértices originales:   ", len(departamentos.get_coordinates()))
print("Vértices simplificados:", len(livianos.get_coordinates()))
```

### ▶️ Paso 2 — El mapa interactivo más simple

Igual que con `.plot()`, empezamos con lo mínimo: el método y el nombre de la columna. Probá
ampliar con la rueda del mouse y pasar el cursor por encima de un departamento.

```python
livianos.explore(column="porcentaje_nbi")
```

Funciona, y ya muestra algo que el mapa estático no podía: al pasar el mouse aparece una etiqueta
con **todos** los datos de ese departamento. Pero son demasiados, la paleta es la de por defecto y
la escala es continua.

### ▶️ Paso 3 — Elegir qué muestra la etiqueta

`tooltip` recibe la lista de columnas que queremos ver, y `tooltip_kwds` con `aliases` les pone un
nombre legible, en el mismo orden.

```python
livianos.explore(
    column="porcentaje_nbi",
    tooltip=["departamento", "provincia", "hogares", "hogares_nbi", "porcentaje_nbi"], #Columnas visibles
    tooltip_kwds={"aliases": ["Departamento", "Provincia", "Hogares", # Alias de las columnas visibles (mismo orden)
                              "Hogares con NBI", "% con NBI"]},
    popup=False,
)
```

### ▶️ Paso 4 — La misma cartografía que en el mapa estático

Ahora le pasamos las mismas decisiones que tomamos antes: paleta secuencial, clasificación por
quiebres naturales, cinco clases, leyenda con título, y un mapa base gris claro para que no
compita con los colores del dato.

Guardamos el resultado en una variable, `mapa`, porque en el paso siguiente le vamos a agregar una
capa.

```python
mapa = livianos.explore(
    column="porcentaje_nbi",
    cmap="Reds",
    scheme="FisherJenks",
    k=5,
    tiles="CartoDB.Positron",
    tooltip=["departamento", "provincia", "porcentaje_nbi"],
    tooltip_kwds={"aliases": ["Departamento", "Provincia", "% con NBI"]},
    popup=False,
    legend_kwds={"caption": "% de hogares con NBI"},
    name="% de hogares con NBI",
)

mapa
```

### ▶️ Paso 5 — Agregar una capa y el control de capas

Un mapa interactivo puede tener varias capas que el lector prende y apaga. Le agregamos el contorno
de las provincias: se dibuja **sobre el mapa que ya existe** pasándolo en el parámetro `m`, y
`style_kwds` define el estilo —línea negra, sin relleno—.

`folium.LayerControl()` agrega el cuadrito de arriba a la derecha para prender y apagar capas.

```python
provincias_livianas.explore(
    m=mapa,                     # dibujar sobre el mapa anterior
    style_kwds={"fill": False, "color": "black", "weight": 1},
    tooltip=["provincia"],
    name="Provincias",
)

folium.LayerControl().add_to(mapa)

mapa
```

### 🔍 Interpretación — para qué sirve y para qué no

El mapa interactivo resolvió el problema del mapa estático: ahora se puede buscar un departamento,
leer su valor exacto y compararlo con el vecino. Para **explorar** los datos y para revisar un
mapa antes de publicarlo.

Pero tiene costos que conviene conocer:

- **No es reproducible en papel** ni en un PDF, y muchos trabajos académicos se entregan así.
- **Invita a mirar el dato de cada unidad** y no el patrón general, que es justamente lo que un
  mapa temático viene a mostrar.
- **El mapa base puede competir** con el dato: si el fondo tiene calles, nombres y colores, el
  lector ya no distingue qué es la variable.

## 11. Mapas de calor: el mapa que casi siempre está mal

### 🧭 Concepto

Un **mapa de calor** (*heatmap*) convierte un conjunto de puntos en una **superficie continua** de
colores: donde hay muchos puntos cerca, la superficie se pinta "caliente". Técnicamente es una
**estimación de densidad**: alrededor de cada punto se reparte un poco de valor y las
contribuciones se suman.

Es el tipo de mapa más popular de los últimos años y el que más se usa mal. Tiene **un** caso de
uso legítimo: **eventos puntuales individuales**, muchos y localizados uno por uno. Delitos,
llamados al 911, focos de incendio, tuits, accidentes de tránsito.

Vamos a hacer primero el mapa de calor que **no** hay que hacer —el que aparece en la mayoría de
las notebooks— y después el caso correcto.

### ▶️ Paso 1 — El mapa de calor mal hecho

Tomamos los centroides de los departamentos y los ponderamos por la cantidad de hogares con NBI.
`HeatMap` de Folium necesita una lista de `[latitud, longitud, peso]`.

Folium trabaja siempre en coordenadas geográficas, así que antes llevamos los centroides
del bloque 9 a EPSG:4326, y de cada punto tomamos la latitud (`y`) y la longitud (`x`).

Los tres parámetros que controlan el dibujo:

- `radius`: cuánto se expande cada punto, en píxeles.
- `blur`: cuánto se suaviza el borde.
- `max_zoom`: hasta qué nivel de ampliación se recalcula la superficie.

```python
from folium.plugins import HeatMap

puntos_ll = puntos.to_crs("EPSG:4326")   # Folium trabaja en coordenadas geográficas

datos_calor = []
for i, fila in puntos_ll.iterrows():
    datos_calor.append([fila.geometry.y, fila.geometry.x, fila["hogares_nbi"]])

mapa_calor = folium.Map(location=[-35, -65], zoom_start=4, tiles="CartoDB positron")
HeatMap(datos_calor, radius=20, blur=15, max_zoom=4).add_to(mapa_calor)

print(f"{len(datos_calor)} puntos, uno por departamento")
mapa_calor
```

### 🔍 ¿Qué está mal en este mapa?

El mapa se ve muy bien, es el más vistoso de la clase, y **no se puede interpretar**. Tres
razones:

1. **Los puntos no son casos, son centroides.** Cada punto representa un departamento entero, a
   veces de decenas de miles de km². La mancha caliente aparece donde cae el centroide, que puede
   ser un lugar sin población: es el problema de la localización representada de la Clase 2, ahora
   amplificado.
2. **El resultado depende de los parámetros, no del dato.** Cambiá `radius` de 20 a 40 y la mancha
   se duplica. Ningún valor es "el correcto", y el mapa no declara cuál usó.
3. **No hay leyenda posible.** ¿Qué significa el rojo? No son hogares, ni hogares por km²: es un
   número sin unidad que sale del cálculo de densidad. Un mapa temático sin leyenda no es un mapa,
   es una ilustración.

A eso se suma que un departamento grande aporta un solo punto igual que uno chico, de modo que las
zonas con muchas unidades administrativas chicas se sobrerrepresentan por partida doble.

Probá cambiar `radius` a 40 en la celda anterior y volver a ejecutarla.

### ▶️ Paso 2 — El caso en que sí corresponde

Un mapa de calor necesita **puntos que sean casos individuales**. Tenemos unos: las escuelas
primarias de la Clase 1, 22.753 puntos, cada uno una escuela real en su ubicación real.

Acá sí tiene sentido preguntar dónde se concentran, porque cada punto es una unidad equivalente a
las demás y está donde está.

```python
escuelas = gpd.read_file(DATOS + "escuelas_primarias.gpkg")

print(f"{len(escuelas)} escuelas primarias")
print("Sistema de coordenadas:", escuelas.crs)
```

```python
datos_escuelas = []
for i, fila in escuelas.iterrows():
    datos_escuelas.append([fila.geometry.y, fila.geometry.x])

mapa_escuelas = folium.Map(location=[-35, -60], zoom_start=5, tiles="CartoDB positron")
HeatMap(datos_escuelas, radius=8, blur=10).add_to(mapa_escuelas)

mapa_escuelas
```

### 🔍 Interpretación

Este mapa sí dice algo: **dónde se concentran las escuelas primarias**, que es donde se concentra
la población. Cada punto es una escuela, no el promedio de un área.

Sigue sin tener leyenda —es la limitación del método—, así que **no se puede usar para comparar**
dos zonas ni para afirmar cuántas escuelas hay en cada lugar. Para eso volvemos a las coropletas y
a los símbolos proporcionales.

La regla, entonces: el mapa de calor sirve para **mostrar dónde hay aglomeración de casos
individuales**, y solo para eso. Si el dato ya viene agregado por unidad administrativa —como el
censo—, el mapa de calor no corresponde.

| Qué variable visual | Valor (claro-oscuro), tamaño | ¿El dato es una cantidad o una categoría? |
| Qué número va en el mapa | Conteo, porcentaje, densidad | ¿Qué quiero saber: dónde hay más, dónde es más probable o dónde se concentra? |
| Qué tipo de mapa | Coropletas, símbolos proporcionales, mapa de calor | ¿El dato es relativo, absoluto, o son casos individuales? |
| Dónde se cortan las clases | Intervalos iguales, cuantiles, quiebres naturales | ¿Estoy mostrando la distribución o fabricando una? |
| Cómo se escala el símbolo | Por área | ¿Lo que el ojo compara es proporcional al dato? |
| Estático o interactivo | `.plot()`, `.explore()` | ¿Es para explorar o para publicar? |

### Glosario de la clase

| Término | Definición |
|---|---|
| **Mapa temático** | Mapa que representa la distribución de una variable. Se distingue del mapa base, que describe el territorio. |
| **Variable visual** | Propiedad gráfica que puede variar para transmitir un dato: tamaño, valor, grano, color, orientación, forma (Bertin). |
| **Valor** (variable visual) | Claro-oscuro. Transmite orden, de poco a mucho. |
| **Dato absoluto** | Un total o conteo. Depende del tamaño de la unidad. |
| **Dato relativo** | Una razón: porcentaje, tasa o densidad. Permite comparar unidades distintas. |
| **Normalizar** | Dividir un conteo por su población de referencia para obtener un dato relativo. |
| **Mapa de coropletas** | Mapa que pinta cada unidad areal según su valor, agrupado en clases. Solo para datos relativos. |
| **Corograma** | Cada unidad areal de un mapa de coropletas. |
| **Clasificación** | Agrupamiento de los valores en clases para asignarles un color. |
| **Intervalos iguales** | Clases del mismo ancho entre el mínimo y el máximo. |
| **Cuantiles** | Clases con la misma cantidad de unidades. |
| **Quiebres naturales (Fisher-Jenks)** | Clases que minimizan la diferencia de valores dentro de cada una. |
| **Paleta secuencial** | Colores de claro a oscuro, para una variable que va de menos a más. |
| **Símbolos proporcionales** | Círculos cuya **área** es proporcional al valor. |
| **Mapa de calor** | Superficie de densidad estimada a partir de puntos. Solo válido con casos individuales. |

### Lo que quedó pendiente

- Todos los mapas de hoy usan el **departamento** como unidad. Adentro de un departamento la
  pobreza no es uniforme, y en la Clase 2 vimos cómo la unidad elegida cambia el resultado. ¿Cómo
  se vería el porcentaje de NBI por radio censal?
- Vimos tres tipos de mapa temático y el mapa de calor. Quedaron nombrados los **mapas de
  puntos**, las **isolíneas**, los **cartogramas** y los **dasimétricos**; los dos últimos vuelven
  en la Clase 8.
- El **color** da para mucho más: paletas divergentes para variables con un centro, cuántos tonos
  distingue el ojo, qué pasa con el daltonismo y con la impresión en blanco y negro. Está en
  *Diseño Cartográfico*, caps. V y VI.
- Hoy **mostramos** patrones. Pero "el norte se ve más oscuro" es una impresión visual. ¿Cómo se
  **mide** si la pobreza está realmente agrupada en el espacio? → **Clase 7**.
